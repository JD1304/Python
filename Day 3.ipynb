```python
# Variable length arguments:

# Sometimes we can pass variable number of arguments to our
# function, such type of
# arguments are called variable length arguments.

# We can declare a variable length argument with * symbol as
# follows

# def f1(*n):

# We can call this function by passing any number of arguments
# including zero number.
# Internally all these values represented in the form of tuple.
def add(*a):
  for x in a:
    print(x, end = " ")

add(1, 2, 3, 4)
def sum(*n):
  total = 0
  for n1 in n:
    total = total + n1
  print("The sum is: ", total)
print()
sum(1,2,2,3,4,5,)
sum(1,2,2,34,4,5,6,6,6,7,8,)
```

```python
#Recurssive Func: Which call itself
def factorial(n):
  if n == 0 or n == 1:
    print(1)
    return 1
  else:
    print(f"{n} * ", end = "")
    return n * factorial(n-1)
n = int(input("Enter any Number: "))
res = factorial(n)
print(res)
```

```python
def mul(x, y):
  if y == 1:
    return x
  elif x==1:
    return x
  elif x==0 or y==0:
    return 0
  else:
    return x+mul(x,y-1)
mul(3,5)
```
