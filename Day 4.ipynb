```python
class TooYoungException(Exception):
  def __init__(self, arg):
    self.msg = arg


class TooOldException(Exception):
  def __init__(self, arg):
    self.msg = arg


age = int(input("Enter your age: "))
if age > 60:
    raise TooYoungException("Wait for some time we will find your best match")
elif age < 18:
    raise TooOldException("You are too young")
else:
    print("You are eligible to marry")
```

```python
arr = [16,17,4,3,5,2]
l = []
for i in range(len(arr)):
    maxx = arr[i]
    for j in range(i,len(arr)):
        if(arr[j]>maxx):
            maxx = arr[j]
    l.append(maxx)
print(l)
```

```python
list1 = [1,2,3,2,1]
count = 0
for i in range(len(list1)):
    if list1[i]==list1[-(i+1)]:
      count+=1
    else:
      count = 0
if count > 0:
  print("Palindrome")
else:
  print("Not Palindrome")
```

```python
# merge 2 sorted arrays
a1 = [1,3,5]
a2 = [2,4,6]
ans = []
i=0
j=0
f = 0
while(True):
    if(i>len(a1)-1):
        f = 1
        break

    elif(j>len(a2)-1):
        f = 2
        break

    if a1[i]<a2[j]:
        ans.append(a1[i])
        i=i+1
    else:
        ans.append(a2[j])
        j=j+1

if f ==1:
    for i in range(j,len(a2)):
        ans.append(a2[i])
else:
    for i in range(i,len(a1)):
        ans.append(a1[i])
ans
```

```python
#Instance if any data or variable is created inside a memory it is called as instance
#Constructor it is used to initialize object's state
# it is __init__(self)
class Student:
  def __init__(self, name, age, percentage):
    print("Constructor is called....")
    self.name = name
    self.age = age
    self.percentage = percentage
  def showInfo(self):
    print("Hello ", self.name)
    print("Your Age: ", self.age)
    print("Your Percentage: ", self.percentage)

s1 = Student("Jash", 20, 88)
s1.showInfo()

```
